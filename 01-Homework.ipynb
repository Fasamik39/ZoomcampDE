```python
time = pd.read_csv("./taxi_zone_lookup.csv")
time.head()
```

# Q3

```python
data[(data["lpep_pickup_datetime"] >= "2025-11-01") & (data["lpep_pickup_datetime"] < "2025-12-01") & (data["trip_distance"] <= 1)].drop_duplicates().shape[0]
```

# Q4

```python
data[data["trip_distance"] < 100].sort_values("trip_distance", ascending=False)
```

# Q5

```python
(
    data[(data["lpep_pickup_datetime"] >= "2025-11-18") & (data["lpep_dropoff_datetime"] < "2025-11-19")]
    .merge(time)
    .groupby("Zone")
    .agg({
        "total_amount": "sum"
    })
    .reset_index()
    .sort_values("total_amount", ascending=False)
)
```

# Q6

```python
merged = data.merge(time, left_on="PULocationID", right_on="LocationID", how="left")
merged = merged.rename(columns={
    "Zone": "pickup_zone"
})
merged = merged.drop(columns=["LocationID", "Borough", "service_zone"])

merged = merged.merge(time, left_on="DOLocationID", right_on="LocationID", how="left")
merged = merged.rename(columns={
    "Zone": "dropoff_zone"
})
merged = merged.drop(columns=["LocationID", "Borough", "service_zone"])

merged.head()
```

```python
(
    merged[(merged["lpep_pickup_datetime"] >= "2025-11-01") & (merged["lpep_dropoff_datetime"] < "2025-12-01") & (merged["pickup_zone"] == "East Harlem North")]
    .sort_values("tip_amount", ascending=False)
        
)
```
